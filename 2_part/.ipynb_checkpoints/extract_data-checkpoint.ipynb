```python
import pandas as pd
```

```python
import re

# Read the file
with open('./classical_cancer.txt', 'r') as file:
    content = file.read()

# Regex pattern to match the line and capture the number
pattern = r"Accuracy Classic-NN: (\d+\.\d+)"
matches = re.findall(pattern, content)

clas_NN_Cancer = {'Acc': []}

# Print the extracted values
for match in matches:
    clas_NN_Cancer['Acc'].append(match)
```

```python
pd.DataFrame(clas_NN_Cancer)
```
