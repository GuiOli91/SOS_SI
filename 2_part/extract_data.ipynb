```python
import pandas as pd
```

```python
import re

# Read the file
with open('./classical_cancer.txt', 'r') as file:
    content = file.read()

# Regex pattern to match the line and capture the number
pattern = r"Accuracy Classic-NN: (\d+\.\d+)"
matches = re.findall(pattern, content)

clas_NN_Cancer = {'Acc': []}

# Print the extracted values
for match in matches:
    clas_NN_Cancer['Acc'].append(match)
```

```python
pd.DataFrame(clas_NN_Cancer)
```

```python
with open('./classical_iris.txt', 'r') as file:
    content = file.read()

# Regex pattern to match the line and capture the number
pattern = r"Accuracy Classic-NN: (\d+\.\d+)"
matches = re.findall(pattern, content)

clas_NN_Iris = {'Acc': []}

# Print the extracted values
for match in matches:
    clas_NN_Iris['Acc'].append(match)
```

```python
pd.DataFrame(clas_NN_Iris)
```

```python
# Read the file
with open('./output.txt', 'r') as file:
    content = file.read()

# Regex pattern to match the line and capture the number
pattern = r"PSO parameters C1:  (\d+(?:\.\d+)?) C2:  (\d+(?:\.\d+)?) W:  (\d+(?:\.\d+)?) Swarmsize:  (\d+) Ieteration:  (\d+)\n\n\nAccuracy PSO-NN: (\d+(?:\.\d+)?)"
matches = re.findall(pattern, content)

# Print the extracted values
df_PSO_Cancer = pd.DataFrame(matches, columns=['C1', 'C2', 'W', 'Swarmsize', 'Iteration', "Accuracy"])
```

```python
# Assuming df_PSO_Cancer is your DataFrame
df_PSO_Cancer = df_PSO_Cancer.apply(pd.to_numeric, errors='coerce')

# Check for any NaN values after conversion
print(df_PSO_Cancer.isna().sum())
```

```python
df_PSO_Cancer.groupby('Swarmsize')['Accuracy'].agg(['mean', 'median', 'std'])
```

```python
df_PSO_Cancer.groupby('W')['Accuracy'].agg(['mean', 'median', 'std'])
```

```python
df_PSO_Cancer.groupby('C2')['Accuracy'].agg(['mean', 'median', 'std'])
```

```python
df_PSO_Cancer.groupby('C1')['Accuracy'].agg(['mean', 'median', 'std'])
```

```python
# Read the file
with open('./PSO-Iris.txt', 'r') as file:
    content = file.read()

# Regex pattern to match the line and capture the number
pattern = r"PSO parameters C1:  (\d+(?:\.\d+)?) C2:  (\d+(?:\.\d+)?) W:  (\d+(?:\.\d+)?) Swarmsize:  (\d+) Ieteration:  (\d+)\n\n\nAccuracy PSO-NN: (\d+(?:\.\d+)?)"
matches = re.findall(pattern, content)

# Print the extracted values
df_PSO_Iris = pd.DataFrame(matches, columns=['C1', 'C2', 'W', 'Swarmsize', 'Iteration', "Accuracy"])
```

```python
df_PSO_Iris.sort_values(by=['Accuracy', 'Swarmsize', 'C1', 'C2', 'W'], ascending=[False, True, True, True, True], inplace=True)
```

```python
df_PSO_Cancer.sort_values(by=['Accuracy', 'Swarmsize', 'C1', 'C2', 'W'], ascending=[False, True, True, True, True], inplace=True)
```

```python
# Assuming df_PSO_Cancer is your DataFrame
df_PSO_Iris = df_PSO_Iris.apply(pd.to_numeric, errors='coerce')

# Check for any NaN values after conversion
print(df_PSO_Iris.isna().sum())
```

```python
df_PSO_Iris.groupby('Swarmsize')['Accuracy'].agg(['mean', 'median', 'std'])
```

```python
df_PSO_Iris.groupby('W')['Accuracy'].agg(['mean', 'median', 'std'])
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming clas_NN_Cancer, clas_NN_Iris, df_PSO_Cancer, df_PSO_Iris are defined in your environment

# Convert accuracies to float
clas_cancer_acc = [float(x) for x in clas_NN_Cancer['Acc']]
clas_iris_acc = [float(x) for x in clas_NN_Iris['Acc']]

# Create combined DataFrame
data = []
for acc in clas_cancer_acc:
    data.append({'Accuracy': acc, 'Dataset': 'Cancer', 'Type': 'Classic'})
for acc in clas_iris_acc:
    data.append({'Accuracy': acc, 'Dataset': 'Iris', 'Type': 'Classic'})
for acc in df_PSO_Cancer['Accuracy']:
    data.append({'Accuracy': acc, 'Dataset': 'Cancer', 'Type': 'PSO'})
for acc in df_PSO_Iris['Accuracy']:
    data.append({'Accuracy': acc, 'Dataset': 'Iris', 'Type': 'PSO'})

df_combined = pd.DataFrame(data)

# Create boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(data=df_combined, x='Dataset', y='Accuracy', hue='Type')
plt.title('Accuracies for Classic NN and PSO NN',
         fontweight='bold')
plt.ylim(0, 1.1)  # Set y-axis limits from 0 to 1
# plt.gca().invert_yaxis()  # Invert the y-axis
plt.grid(True, linestyle='--')
plt.savefig('boxplot_models.png')
plt.show()
```
